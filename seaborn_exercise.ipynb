```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

dates = pd.date_range('2021-01-01',periods=100)

prices = pd.Series(range(100)) + np.random.randn(100).cumsum()
stock_data  = pd.DataFrame({
    'Date': dates, 'Price': prices
})

sns.lineplot(x='Date', y='Price', data=stock_data)

plt.title('Stock Price Over Time')
plt.xlabel('Date')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
tips = sns.load_dataset('tips')

# 막대 그래프 그리기
sns.barplot(x='day', y='tip', data=tips, estimator=sum)
plt.title('Total Tips by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Tip Amount')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 데이터 생성 (회사의 월별 제품 판매량)
data = {
    'January': [500, 600, 700, 400],
    'February': [550, 620, 680, 450],
    'March': [530, 610, 670, 420],
    'April': [580, 640, 710, 460],
    'May': [590, 660, 720, 470],
    'June': [620, 680, 750, 490]
}
companies = ['Company A', 'Company B', 'Company C', 'Company D']

df = pd.DataFrame(data,index=companies)
plt.figure(figsize=(8,6))
sns.heatmap(df,annot=True,cmap='YlGnBu',linewidths=0.5)

plt.show()
```

```python
# 데이터 조작: 다양한 상관관계 (1, -1, 0에 가까운 값이 나오도록)
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 수정된 데이터 (다양한 상관관계 유도)
data = {
    'Stock Price': [120, 135, 150, 160, 145, 170, 185, 190],    # 주가
    'Revenue': [50, 55, 65, 70, 67, 75, 80, 85],                # 매출
    'Profit': [30, 28, 25, 22, 21, 20, 18, 15],                # 이익 (감소하는 값)
    'Expenses': [60, 58, 55, 53, 50, 48, 45, 43],              # 비용 (감소)
    'Debt': [80, 75, 70, 68, 72, 74, 78, 82],                  # 부채 (변동)
    'R&D Spending': [10, 10, 15, 15, 20, 20, 25, 25]           # 연구개발비 (일정한 패턴)
}

# 데이터프레임 생성
df = pd.DataFrame(data)

# 상관계수 계산
corr_matrix = df.corr()

# Heatmap 그리기
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)

# 타이틀 추가
plt.title('Correlation Between Financial Metrics')

# 그래프 보여주기
plt.show()

# 상관계수 출력
corr_matrix
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 예시 데이터 생성
data = {
    'Class': ['A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'B', 'C', 'C', 'C', 'C', 'C'],
    'Score': [85, 88, 90, 93, 95, 70, 75, 80, 82, 85, 55, 60, 65, 70, 75]
}

# Violin Plot 그리기
plt.figure(figsize=(8, 6))
sns.violinplot(x='Class', y='Score', data=data)

# 타이틀 추가
plt.title('Distribution of Exam Scores by Class')

# 그래프 보여주기
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 예시 데이터 생성
data = {
    'Class': ['A', 'A', 'A', 'B', 'B', 'B'],
    'Score': [85, 88, 90, 70, 75, 80],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}

# Violin Plot 그리기
plt.figure(figsize=(8, 6))
sns.violinplot(x='Class', y='Score', hue='Gender', data=data, split=True)

# 타이틀 추가
plt.title('Distribution of Exam Scores by Gender and Class')

# 그래프 보여주기
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

# 아이리스 데이터셋 로드
iris = load_iris()
df_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
df_iris['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# Pair Plot 그리기
sns.pairplot(df_iris, hue='species')

# 그래프 보여주기
plt.show()
```

```python
# 예시: 회귀선을 포함한 Pair Plot
sns.pairplot(df_iris, hue='species', kind='reg')
plt.show()
```
