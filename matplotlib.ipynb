````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# 사용자 정의 컬러맵 설정
# LinearSegmentedColormap.from_list()를 사용하여 색상 리스트를 기반으로 컬러맵 정의
colors = ["#B5B5C2", "#FFF700", "#E31919"]  # 파란색에서 녹색, 빨간색으로 변환
custom_cmap = LinearSegmentedColormap.from_list('custom', colors)

# 2D 배열 데이터 생성
data = np.random.rand(10, 10)

# 사용자 정의 컬러맵 적용
plt.imshow(data, cmap=custom_cmap)
plt.colorbar()

# 타이틀 추가
plt.title('Custom Colormap')

# 그래프 보여주기
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

months = ['January', 'February', 'March', 'April', 'May', 'June', 
            'July', 'August', 'September', 'October', 'November', 'December']
avg_temperatures = [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0]
rainfall = [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
math_scores = [85, 90, 88, 76, 95, 89, 72, 84, 91, 87]
english_scores = [78, 82, 85, 80, 88, 90, 84, 79, 85, 83]
````

````python
plt.figure(figsize=(18,12))
plt.rcParams['font.family']='Malgun Gothic'
# 1. 월별 평균 기온을 시각화하는 선 그래프를 그려보세요.
plt.subplot(3,2,1)
plt.plot(months,avg_temperatures, linestyle="--", marker="^")
plt.title("Monthly Average Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.grid()
plt.xticks(rotation=45)


# 2. 월별 강수량을 시각화하는 막대 그래프를 그려보세요.
plt.subplot(3,2,2)
plt.bar(months,rainfall, color="#E610A9")
plt.title("Monthly Rainfall")
plt.xlabel("Month")
plt.ylabel("Rainfall")
plt.xticks(rotation=45)

# 3. 수학과 영어 점수의 분포를 비교하는 박스플롯 그래프를 그려보세요.
plt.subplot(3,2,3)
plt.boxplot([math_scores, english_scores]
            , labels=["Math","English"])
plt.title("수학 영어 점수 비교")
plt.ylabel("점수")
plt.legend()

# 4. 수학 점수 분포를 분석하는 히스토그램을 그려보세요.
plt.subplot(3,2,4)
plt.hist(math_scores, bins=7, color="blue", edgecolor="black")
plt.title("Math Score")
plt.xlabel("Math")

# 5. 수학 점수와 영어 점수의 상관관계를 분석하는 산점도를 그려보세요.
plt.subplot(3,2,5)
plt.scatter(math_scores,english_scores,color="#F25912")
plt.title("수학 영어 점수 상관관계")
plt.xlabel("Math")
plt.ylabel("English")
# 6. 1~5의 모든 그래프를 하나의 Figure에 배치해 보세요.

plt.tight_layout()
plt.show()
````
